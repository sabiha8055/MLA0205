```python
import csv

data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'Yes']
]

with open("training_data.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(data)

with open("training_data.csv", "r") as f:
    reader = csv.reader(f)
    data = list(reader)

S = ['0'] * 6
G = [['?'] * 6]

for row in data:
    attributes = row[:-1]
    target = row[-1]

    if target == 'Yes':
        G = [g for g in G if all(
            g[i] == '?' or g[i] == attributes[i]
            for i in range(6)
        )]

        for i in range(6):
            if S[i] == '0':
                S[i] = attributes[i]
            elif S[i] != attributes[i]:
                S[i] = '?'

    else:
        new_G = []
        for g in G:
            for i in range(6):
                if g[i] == '?':
                    if S[i] != '?' and S[i] != '0':
                        new_h = g.copy()
                        new_h[i] = S[i]
                        if new_h not in new_G:
                            new_G.append(new_h)
        G = new_G

print("Specific Boundary S:")
print(S)

print("\nGeneral Boundary G:")
for g in G:
    print(g)
```
