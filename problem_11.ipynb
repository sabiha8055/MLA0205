```python

import pandas as pd
from sklearn.tree import DecisionTreeClassifier

# Create dataset
data = {
    'Income': [25000, 50000, 30000, 80000, 60000, 20000],
    'Age': [22, 35, 25, 45, 40, 21],
    'Debt': [20000, 10000, 25000, 5000, 12000, 30000],
    'Credit_Score': ['Low', 'High', 'Low', 'High', 'High', 'Low']
}

df = pd.DataFrame(data)

# Input and output
X = df[['Income', 'Age', 'Debt']]
y = df['Credit_Score']

# Create model
model = DecisionTreeClassifier(random_state=0)

# Train model
model.fit(X, y)

# Predict credit score
prediction = model.predict([[55000, 32, 10000]])

print("Predicted Credit Score:", prediction[0])
```
