```python
import numpy as np

# Sigmoid activation function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Derivative of sigmoid
def sigmoid_derivative(x):
    return x * (1 - x)

# XOR dataset
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
], dtype=float)

y = np.array([
    [0],
    [1],
    [1],
    [0]
], dtype=float)

# Set random seed for reproducible results
np.random.seed(42)

# Initialize weights and biases
input_neurons = 2
hidden_neurons = 4
output_neurons = 1

W1 = np.random.uniform(-1, 1, (input_neurons, hidden_neurons))
b1 = np.zeros((1, hidden_neurons))

W2 = np.random.uniform(-1, 1, (hidden_neurons, output_neurons))
b2 = np.zeros((1, output_neurons))

learning_rate = 0.5
epochs = 20000

# Training using backpropagation
for epoch in range(epochs):
    # Forward propagation
    hidden_input = np.dot(X, W1) + b1
    hidden_output = sigmoid(hidden_input)

    final_input = np.dot(hidden_output, W2) + b2
    predicted_output = sigmoid(final_input)

    # Calculate error
    error = y - predicted_output

    # Backpropagation
    output_delta = error * sigmoid_derivative(predicted_output)
    hidden_error = np.dot(output_delta, W2.T)
    hidden_delta = hidden_error * sigmoid_derivative(hidden_output)

    # Update weights and biases
    W2 += learning_rate * np.dot(hidden_output.T, output_delta)
    b2 += learning_rate * np.sum(output_delta, axis=0, keepdims=True)

    W1 += learning_rate * np.dot(X.T, hidden_delta)
    b1 += learning_rate * np.sum(hidden_delta, axis=0, keepdims=True)

# Test the trained network
hidden_output = sigmoid(np.dot(X, W1) + b1)
predictions = sigmoid(np.dot(hidden_output, W2) + b2)

print("Artificial Neural Network using Backpropagation")
print("\nInput\t\tTarget\tPredicted")
for i in range(len(X)):
    print(
        X[i].astype(int),
        "\t",
        int(y[i, 0]),
        "\t",
        round(float(predictions[i, 0]), 4)
    )

print("\nClassification Results:")
print((predictions >= 0.5).astype(int).flatten())
```
