```python

import numpy as np
from sklearn.mixture import GaussianMixture

# Input data
X = np.array([[1], [2], [3], [8], [9], [10]])

# Create EM model
model = GaussianMixture(n_components=2, random_state=0)

# Train model
model.fit(X)

# Predict clusters
labels = model.predict(X)

print("Cluster Labels:")
print(labels)

print("Cluster Means:")
print(model.means_)
```
